```python
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

print("Random seed:", SEED)
```

```python
import glob

zip_files = glob.glob(
    PROJECT_PATH + "/*.zip"
)

print("Available ZIP files:")

for file in zip_files:
    print(file)
```

```python
import zipfile

if len(zip_files) == 1:

    with zipfile.ZipFile(zip_files[0], 'r') as zip_ref:
        zip_ref.extractall(
            PROJECT_PATH + "/dataset"
        )

    print("Dataset extracted successfully")

else:
    print("Ensure exactly one dataset ZIP is present")
```

```python
DATASET_PATH = PROJECT_PATH + "/dataset"

for root, dirs, files in os.walk(DATASET_PATH):

    level = root.replace(
        DATASET_PATH, ''
    ).count(os.sep)

    if level <= 4:

        indent = " " * 4 * level

        print(
            f"{indent}{os.path.basename(root)}/"
            f" ({len(files)} files)"
        )
```
